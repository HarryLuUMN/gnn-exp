# GNNVisualizer multi-instance minimal test

This notebook creates two `GNNVisualizer` instances in one output area so the multi-instance rendering bug can be checked quickly.

```python
import sys
from pathlib import Path

repo_root = Path.cwd().resolve().parent if Path.cwd().name == "notebooks" else Path.cwd().resolve()
src_path = repo_root / "src"
if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))

from IPython.display import display
from gnn_exp import GNNVisualizer
```

```python
graph_data = {
    "x": [[1.0, 0.0], [0.0, 1.0], [1.0, 1.0]],
    "edge_index": [[0, 1, 1, 2], [1, 0, 2, 1]],
    "y": [0, 1, 0],
}

intm_data = {
    "act0": [[1.0, 0.0], [0.0, 1.0], [1.0, 1.0]],
    "act1": [[0.9, 0.1], [0.2, 0.8], [0.7, 0.3]],
    "softmax": [[0.95, 0.05], [0.10, 0.90], [0.75, 0.25]],
}

model_info = {
    "conv1": {
        "type": "GCNConv",
        "weight": [[1.0, 0.0], [0.0, 1.0]],
        "bias": [0.0, 0.0],
    },
    "classifier": {
        "type": "Linear",
        "weight": [[1.0, 0.0], [0.0, 1.0]],
        "bias": [0.0, 0.0],
    },
}
```

```python
vis_a = GNNVisualizer(
    graphData=graph_data,
    intmData=intm_data,
    modelInfo=model_info,
    subgraphSample=False,
    queries=[[0, 1]],
    mode="node",
)

vis_b = GNNVisualizer(
    graphData=graph_data,
    intmData=intm_data,
    modelInfo=model_info,
    subgraphSample=False,
    queries=[[1, 2]],
    mode="node",
)

display(vis_a)
```
